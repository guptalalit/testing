```python
import os
import json
from ragas.metrics import Faithfulness, AnswerRelevancy
from ragas.llms import BaseRagasLLM
from google import genai
from google.genai.types import GenerateContentConfig
from dotenv import load_dotenv

load_dotenv()

class GeminiRagasLLM(BaseRagasLLM):
    def __init__(self):
        self.client = genai.GenerativeModel('gemini-pro')
        
    def generate(self, prompt: str) -> str:
        response = self.client.generate_content(prompt)
        return response.text
    
    async def agenerate(self, prompt: str) -> str:
        response = self.client.generate_content(prompt)
        return response.text

# Initialize Gemini LLM wrapper
gemini_llm = GeminiRagasLLM()

# Create custom metrics using Gemini
class GeminiFaithfulness(Faithfulness):
    def __init__(self):
        super().__init__(llm=gemini_llm)

class GeminiAnswerRelevancy(AnswerRelevancy):
    def __init__(self):
        super().__init__(llm=gemini_llm)

# Example usage
if __name__ == "__main__":
    from ragas import evaluate
    from datasets import Dataset
    
    # Create sample data
    sample = {
        "question": ["When was the first super bowl?"],
        "answer": ["The first superbowl was held on Jan 15, 1967"],
        "contexts": [[
            "The First AFL–NFL World Championship Game was an American football game played on January 15, 1967, at the Los Angeles Memorial Coliseum in Los Angeles."
        ]]
    }
    
    # Create metrics
    metrics = [
        GeminiFaithfulness(),
        GeminiAnswerRelevancy()
    ]
    
    # Evaluate
    dataset = Dataset.from_dict(sample)
    scores = evaluate(dataset, metrics=metrics)
    
    print("Faithfulness Score:", scores["faithfulness"])
    print("Answer Relevancy Score:", scores["answer_relevancy"])
```

```python
import os
import faiss
import json
import numpy as np
from sentence_transformers import SentenceTransformer
import google.generativeai as genai

class RAG:
    def __init__(self):
        # Initialize paths and models
        self.index_file = "faiss_index.index"
        self.metadata_file = "metadata.json"
        self.embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
        
        # Configure Gemini API
        self.gemini_api_key = os.getenv("GEMINI_API_KEY")
        genai.configure(api_key=self.gemini_api_key)
        self.gemini_model = genai.GenerativeModel('gemini-pro')

    def retrieve_documents(self, query: str, k: int = 3) -> list:
        """
        Retrieve the top-k most relevant documents for the given query using FAISS.
        """
        index = faiss.read_index(self.index_file)
        
        with open(self.metadata_file, "r", encoding="utf-8") as f:
            metadata = json.load(f)

        query_embedding = self.embedding_model.encode([query], convert_to_tensor=False)
        distances, indices = index.search(query_embedding, k)

        results = []
        for i, idx in enumerate(indices[0]):
            if idx >= 0:
                result = metadata[idx]
                result["distance"] = float(distances[0][i])
                results.append(result)

        return results

    def rag(self, query: str) -> str:
        """
        Generate an answer for the given query using the RAG pipeline.
        """
        context_docs = self.retrieve_documents(query, k=3)
        context = "\n\n".join([doc["text"] for doc in context_docs])
        
        prompt = f"""
        Provide the answer for the given question based on the context.
        Question: {query}
        Context: {context}
        """
        
        response = self.gemini_model.generate_content(prompt)
        return response.text, context

# Initialize the RAG pipeline
rag_tool = RAG()

# Example queries to test
example_queries = [
    "What are the symptoms of lung cancer?",
    "How is lung cancer treated?",
    "What is the difference between adenocarcinoma and squamous cell carcinoma?",
    "What are the stages of lung cancer?",
    "What is the survival rate for lung cancer?"
]

# Test each query
for query in example_queries:
    answer, context = rag_tool.rag(query)
    print(f"Query: {query}\nAnswer: {answer}\nContext: {context}\n")
    print("=" * 50 + "\n")
```
